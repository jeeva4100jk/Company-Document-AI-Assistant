```python
import pypdf
print("PDF Library installed successfully!")
```

```python
from pypdf import PdfReader

pdf_path = "Employee_Leave_Policy.pdf"

reader = PdfReader(pdf_path)

print("Total Pages:", len(reader.pages))
```

```python
full_text = ""

for page in reader.pages:
    text = page.extract_text()
    full_text += text + "\n"

print(full_text)
```

```python
print("Total Characters:", len(full_text))
print("First 500 Characters:")
print(full_text[:500])
```

### Text cleaning

```python
import re
clean_text = re.sub(r'\s+', ' ', full_text).strip()
print(clean_text[:500])
```

```python
print("Before:", len(full_text))
print("After:", len(clean_text))

print(clean_text[:500])
```

### Text validation

```python
print("Annual Leave:", "18 days" in clean_text)
print("Sick Leave:", "10 days" in clean_text)
print("Work From Home:", "Work From Home" in clean_text)
```

```python
import re
from pypdf import PdfReader

# Read PDF
pdf_path = "../data&doc/Employee_Leave_Policy.pdf"
reader = PdfReader(pdf_path)

# Extract text
full_text = ""

for page in reader.pages:
    text = page.extract_text()
    full_text += text + "\n"

# Clean text
clean_text = re.sub(r'\s+', ' ', full_text).strip()

print("PDF extraction and cleaning completed!")
print("Total Characters:", len(clean_text))
```

```python
with open("../data&doc/cleaned_text.txt", "w", encoding="utf-8") as file:
    file.write(clean_text)

print("File saved successfully!")
```
