```python
import requests
```

```python
url = "http://localhost:11434/api/generate"
```

```python
# Step 3: Send a request to Ollama
response = requests.post(url,json={"model": "qwen2.5:3b","prompt": "What is Artificial Intelligence?",
"stream": False})

# Step 4: Convert response into JSON
result = response.json()

print(result["response"])
```
