```python
import numpy as np
```

```python
embeddings = np.load("../data&doc/embeddings.npy")

print("embeddings loaded successfully!")
print("embeddings shate:", embeddings.shape)
```

```python
import faiss
```

```python
dimension = embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)

index.add(embeddings.astype("float32"))

print("FAISS index created successfully!")
print("Total Vectors Stored:", index.ntotal)
```

```python
faiss.write_index(index, "../data&doc/faiss_index.bin")

print("FAISS index saved successfully!")
```

```python
loaded_index = faiss.read_index("../data&doc/faiss_index.bin")

print("FAISS index loaded successfully!")
print("Total Vectors:", loaded_index.ntotal)
```
