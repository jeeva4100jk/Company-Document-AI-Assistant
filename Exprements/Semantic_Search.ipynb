```python
import json
import numpy as np
import faiss
from sentence_transformers import SentenceTransformer

print("Libraries imported successfully!")
```

```python
with open("../data&doc/chunks.json",'r' , encoding = "utf-8") as file:
    chunks = json.load(file)
    
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")

print("Embedding model loaded successfully!")
```

```python
index = faiss.read_index("../data&doc/faiss_index.bin")
```

```python
query = "How many sick leaves can I take?"

query_embedding = model.encode([query])

query_embedding = query_embedding.astype("float32")

print("Query embedding created!")
print("Query Shape:", query_embedding.shape)
```

```python
distance, indices = index.search(query_embedding, k=2)

print("Search Results:")

for i in indices[0]:
    print("\n", chunks[i])
```

```python
print("Distances:", distance)
print("Matched Chunk Indices:", indices)
```

```python
best_match_index = indices[0][0]

best_match_distance = distance[0][0]

best_chunk = chunks[best_match_index]

print("Best Matching Chunk:")
print(best_chunk)

print("\nDistance Score:", best_match_distance)
```

```python
query = "How many annual leaves are available?"

query_embedding = model.encode([query]).astype("float32")

distance, indices = index.search(query_embedding, k=2)

for i in indices[0]:
    print("\nRetrieved Chunk:")
    print(chunks[i])
```
